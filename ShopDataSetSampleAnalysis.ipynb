```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("weather_dataset.csv")
```

```python
hourly = df.groupby("hour")["crowd_level"].mean()

plt.figure(figsize=(10,5))
plt.plot(hourly.index, hourly.values, marker='o')
plt.title("Average Crowd Level by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Crowd Level")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x="temperature", y="crowd_level", hue="weather")
plt.title("Temperature vs Crowd Level")
plt.show()
```

```python
weather_avg = df.groupby("weather")["crowd_level"].mean().sort_values()
plt.figure(figsize=(10,5))
weather_avg.plot(kind="bar")
plt.title("Average Crowd Level by Weather Type")
plt.ylabel("Crowd Level")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
monthly = df.groupby("month")["crowd_level"].mean()

plt.figure(figsize=(10,5))
plt.plot(monthly.index, monthly.values, marker='o')
plt.title("Average Crowd Level by Month")
plt.xlabel("Month")
plt.ylabel("Crowd Level")
plt.grid(True)
plt.show()
```
